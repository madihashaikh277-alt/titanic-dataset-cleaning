# Titanic Dataset Cleaning Project

## Introduction

The objective of this project is to clean a public Titanic dataset using Python and the Pandas library in Google Colab.

The dataset contains information about Titanic passengers, including age, gender, passenger class, fare, and survival status.

In this project, I checked and handled missing values, duplicate records, incorrect data types, and inconsistent values to prepare a clean dataset for further analysis.

**Tools Used:** Python, Pandas, Google Colab.


```python

import pandas as pd
df = pd.read_csv('titanic.csv')
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Cabin'] = df['Cabin'].fillna('Unknown')
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python

print("Duplicate rows:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

```python

# Check data types of all columns
print("Data types before correction:")
print(df.dtypes)
```

```python

print("Unique values in Sex:")
print(df['Sex'].unique())

print("\nUnique values in Embarked:")
print(df['Embarked'].unique())
```

```python

# Save the cleaned dataset
df.to_csv('titanic_cleaned.csv', index=False)
print("Cleaned dataset saved successfully!")
print("Final dataset shape:", df.shape)
```

```python

from google.colab import files
files.download('titanic_cleaned.csv')
print("Database Downloaded Sucessfully")
```

# Conclusion

The Titanic dataset was successfully cleaned using Python and Pandas.

- Missing values were handled using the median, mode, and the value "Unknown".
- No duplicate rows were found.
- No incorrect data types were identified.
- The Sex and Embarked columns were checked for inconsistent values.
- The final dataset contains 891 rows and 12 columns, with zero missing values.

The cleaned dataset was saved as `titanic_cleaned.csv`.
